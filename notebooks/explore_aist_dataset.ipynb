```python
import matplotlib.pyplot as plt
import pickle

with open("annotations/keypoints2d/gBR_sBM_cAll_d04_mBR0_ch01.pkl", "rb") as f:
    data = pickle.load(f)

cam = 0
num_frames = 100

frames = list(range(0, num_frames, 5))

fig, axes = plt.subplots(len(frames), 1, figsize=(6, 90))

for i, frame in enumerate(range(0, num_frames, 5)):

    kp = data["keypoints2d"][cam][frame] #(17,3)

    x = kp[:, 0]
    y = kp[:, 1]

    ax = axes[i]

    ax.scatter(x, y)

    ax.invert_yaxis()
    ax.set_xlim(0, 1920)
    ax.set_ylim(1080, 0)

    ax.set_title(f"Frame {frame}")

plt.tight_layout()
plt.show()
```
